```python
import numpy as np

np.random.seed(42)

# 1. DATA
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],
              [1],
              [1],
              [0]])

# 2. WEIGHTS
W1 = np.random.randn(2,3)
b1 = np.zeros((1,3))

W2 = np.random.randn(3,1)
b2 = np.zeros((1,1))

# activation
def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

# 3–5 TRAINING
for i in range(10000):

    # 3. FORWARD
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)

    # 4. LOSS
    loss = np.mean((y - A2)**2)

    # 5. BACKPROP
    dA2 = -(y - A2)
    dZ2 = dA2 * sigmoid_derivative(A2)

    dW2 = np.dot(A1.T, dZ2)
    db2 = np.sum(dZ2, axis=0)

    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * sigmoid_derivative(A1)

    dW1 = np.dot(X.T, dZ1)
    db1 = np.sum(dZ1, axis=0)

    # UPDATE
    W1 -= 0.1 * dW1
    b1 -= 0.1 * db1
    W2 -= 0.1 * dW2
    b2 -= 0.1 * db2

# OUTPUT
print(A2)
```
